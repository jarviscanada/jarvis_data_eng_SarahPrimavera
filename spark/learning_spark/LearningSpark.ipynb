```
# Retrieve the start times of members' bookings
df = spark.sql("SELECT starttime FROM cd.bookings WHERE memid = (SELECT memid FROM cd.members WHERE surname = 'Farrell' AND firstname = 'David');")
display(df)
```

```
# Work out the start times of bookings for tennis courts
df = spark.sql("SELECT starttime, name FROM cd.bookings INNER JOIN cd.facilities on cd.bookings.facid = cd.facilities.facid WHERE name in ('Tennis Court 2','Tennis Court 1') AND starttime >= '2012-09-21' AND starttime < '2012-09-22' ORDER BY starttime;")
display(df)
```

```
# Produce a list of all members, along with their recommender
df = spark.sql("select mems.firstname, mems.surname, recs.firstname, recs.surname from cd.members mems left outer join cd.members recs on recs.memid = mems.recommendedby order by mems.surname, mems.firstname;")
display(df)
```

```
# Produce a list of all members who have used a tennis court
df = spark.sql("select distinct mems.firstname || ' ' || mems.surname as member, facs.name as facility from cd.members mems inner join cd.bookings bks on mems.memid = bks.memid inner join cd.facilities facs on bks.facid = facs.facid where facs.name in ('Tennis Court 2','Tennis Court 1') order by member, facility;")
display(df)
```

```
# Produce a list of all members, along with their recommender, using no joins.
df = spark.sql("SELECT DISTINCT mems.firstname|| ' ' ||mems.surname as member, (SELECT DISTINCT recs.firstname|| ' ' || recs.surname as recommender FROM cd.members recs WHERE recs.memid = mems.recommendedby) FROM cd.members mems ORDER BY member;")
display(df)
```

```
# Count the number of recommendations each member makes.
df = spark.sql("select recommendedby, count(*) from cd.members where recommendedby is not null group by recommendedby order by recommendedby; ")
display(df)
```

```
# List the total slots booked per facility
df = spark.sql("SELECT facid, sum(slots) FROM cd.bookings GROUP BY facid ORDER BY facid;")
display(df)
```

```
# List the total slots booked per facility in a given month
df = spark.sql("SELECT facid, sum(slots) FROM cd.bookings WHERE starttime >= '2012-09-1' AND starttime < '2012-10-01' GROUP BY facid ORDER BY sum(slots);")
display(df)
```

```
# List the total slots booked per facility per month
df = spark.sql("SELECT facid, extract(month FROM starttime) as month, sum(slots) FROM cd.bookings WHERE extract(year FROM starttime) = 2012 GROUP BY facid, month ORDER BY facid, month;")
display(df)
```

```
# Find the count of members who have made at least one booking
df = spark.sql("SELECT COUNT(DISTINCT memid) FROM cd.bookings;")
display(df)
```

```
# List each member's first booking after September 1st 2012
df = spark.sql("SELECT surname, firstname, cd.members.memid, min(cd.bookings.starttime) as starttime FROM cd.bookings INNER JOIN cd.members ON cd.members.memid = cd.bookings.memid WHERE starttime >= '2012-09-01' GROUP BY surname, firstname, cd.members.memid ORDER BY cd.members.memid;")
display(df)
```

```
# Format the names of members
df = spark.sql("SELECT surname||', ' ||firstname AS name FROM cd.members;")
display(df)
```

```
# Perform a case-insensitive search
df = spark.sql("select * from cd.facilities where upper(name) like 'TENNIS%';")
display(df)
```

```
# Find telephone numbers with parentheses
df = spark.sql("SELECT memid, telephone FROM cd.members WHERE telephone ~ '[()]';")
display(df)
```

```
# Count the number of members whose surname starts with each letter of the alphabet
df = spark.sql("SELECT substr (mems.surname,1,1) AS letter, COUNT(*) AS COUNT FROM cd.members mems GROUP BY letter ORDER BY letter;")
display(df)
```

```
# Generate a list of all the dates in October 2012
df = spark.sql("select generate_series(timestamp '2012-10-01', timestamp '2012-10-31', interval '1 day') as ts;  ")
display(df)
```

```
# Return a count of bookings for each month
df = spark.sql("select date_trunc('month', starttime) as month, count(*) from cd.bookings group by month order by month;")
display(df)
```
